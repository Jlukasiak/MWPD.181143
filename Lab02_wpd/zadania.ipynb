```python
iris = pd.read_csv('iris.data')

iris.head()
```

```python
# --- ZADANIE 3 ---

iris_numeric = iris.select_dtypes(include=['number'])

srednia_globalna = iris_numeric.mean()
print("Średnia globalna dla wszystkich cech:")
print(srednia_globalna)
print("-" * 30)

srednia_grupowana = iris.groupby('class').mean()
print("Średnia dla poszczególnych klas (gatunków):")
print(srednia_grupowana)
```

```python
# --- ZADANIE 4 ---


red_wine = pd.read_csv('winequality-red.csv', sep=';')
white_wine = pd.read_csv('winequality-white.csv', sep=';')

red_wine['color'] = 'red'
white_wine['color'] = 'white'

wine = pd.concat([red_wine, white_wine], ignore_index=True)

features = wine.drop(columns=['quality', 'color'])
variances = features.var().sort_values()

print("Cechy uszeregowane według wariancji (rosnąco):")
print(variances)
print("-" * 30)

var_by_quality = wine.drop(columns='color').groupby('quality').var()
print("Wariancja cech w podziale na jakość (quality):")
display(var_by_quality)

mean_var_per_quality = var_by_quality.mean(axis=1)
print("\nUśredniona wariancja dla każdej klasy jakości:")
print(mean_var_per_quality)

std_by_color = wine.drop(columns='quality').groupby('color').std()
print("\nOdchylenie standardowe w podziale na kolor wina:")
display(std_by_color)
```

```python
# --- ZADANIE 5 ---

features_wine = wine.drop(columns=['quality', 'color'])

kwartyle = features_wine.quantile([0.25, 0.5, 0.75])

print("Wartości kwartyli dla cech w zbiorze Wine Quality:")
display(kwartyle)
```

```python
wine.columns = [col.replace(' ', '_') for col in wine.columns]

print(wine.columns)
```

```python
# --- ZADANIE 6 ---

progu_cukru = wine['residual_sugar'].quantile(0.95)

print(f"Wartość odcinająca 95% wyników (P95): {progu_cukru:.2f}")

najslodsze_wina = wine[wine['residual_sugar'] > progu_cukru]

print(f"Liczba znalezionych pozycji: {len(najslodsze_wina)}")
display(najslodsze_wina)
```
