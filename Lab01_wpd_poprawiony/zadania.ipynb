```python
import numpy as np
import pandas as pd
```

```python
# --- ZADANIE 1 ---
dane = list(range(1, 101))
indeks = [f"ean_{n:03}" for n in range(1, 101)]
zadanie_1 = pd.Series(data=dane, index=indeks)
zadanie_1.head()
```

```python
# --- ZADANIE 2 ---
podstawa = np.arange(1, 11).reshape(10, 1)
wykladniki = np.arange(1, 11)
tablica_np = np.power(podstawa, wykladniki)
nazwy_kolumn = ['base'] + [f'pow_{n}' for n in range(2, 11)]
df_potegi = pd.DataFrame(data=tablica_np, columns=nazwy_kolumn)
df_potegi
```

```python
# --- ZADANIE 3 ---
wspoldzieli = df_potegi.values.base is tablica_np
print(f"Czy df_potegi współdzieli pamięć? {wspoldzieli}")

df_potegi_shared = pd.DataFrame(data=tablica_np, copy=False)
df_potegi_shared.columns = nazwy_kolumn # nadajemy te same nazwy kolumn

# 3. Sprawdzamy ponownie dla nowej ramki
wspoldzieli_nowa = df_potegi_shared.values.base is tablica_np
print(f"Czy df_potegi_shared współdzieli pamięć? {wspoldzieli_nowa}")
```

```python
# --- ZADANIE 4 ---


print("Sytuacja przed optymalizacją:")
print(df_potegi_shared.dtypes)
pamiec_przed = df_potegi_shared.memory_usage(deep=True).sum()
print(f"Pamięć przed: {pamiec_przed} bajtów")


df_zad4 = df_potegi_shared.apply(pd.to_numeric, downcast='integer')


print("\nSytuacja po optymalizacji:")
print(df_zad4.dtypes)
pamiec_po = df_zad4.memory_usage(deep=True).sum()
print(f"Pamięć po: {pamiec_po} bajtów")

# 4. Uzasadnienie
roznica = pamiec_przed - pamiec_po
print(f"\nUzasadnienie: Poprzez zmianę typów (np. z int64 na mniejsze int8 tam, gdzie to możliwe), zaoszczędzono {roznica} bajtów.")
```

```python
# --- ZADANIE 5 ---

iris = pd.read_csv('iris.data')


iris.head()
```

```python
# --- ZADANIE 6 ---

iris.columns = [col.replace(' ', '_') for col in iris.columns]

iris.head(10)
```

```python
# --- ZADANIE 7 ---


mem_shallow = iris.memory_usage().sum()
mem_deep = iris.memory_usage(deep=True).sum()

print(f"Pamięć (bez deep): {mem_shallow} bajtów")
print(f"Pamięć (z deep=True): {mem_deep} bajtów")
print("-" * 30)

float_cols = iris.select_dtypes(include=['float64']).columns
print(f"Kolumny do zmiany: {list(float_cols)}")

iris_smaller = iris.copy()
iris_smaller[float_cols] = iris_smaller[float_cols].astype('float32')

mem_smaller_deep = iris_smaller.memory_usage(deep=True).sum()

print(f"Typy danych w iris_smaller:\n{iris_smaller.dtypes}")
print(f"\nNowa pamięć (deep=True): {mem_smaller_deep} bajtów")

oszczednosc = mem_deep - mem_smaller_deep
print(f"\nUzasadnienie: Zmiana typu z float64 na float32 zmniejszyła wagę ramki o {oszczednosc} bajtów.")
```
